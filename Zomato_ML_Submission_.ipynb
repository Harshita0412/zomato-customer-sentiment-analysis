## 3. ***Data Wrangling***

### Data Wrangling Code

````python

# Create Working Copies for Data Wrangling

metadata_clean = metadata_df.copy()
reviews_clean = reviews_df.copy()

print("Metadata working copy:", metadata_clean.shape)
print("Reviews working copy:", reviews_clean.shape)
````

````python

# Convert Restaurant Cost into Numeric Format

metadata_clean['Cost'] = (
    metadata_clean['Cost']
    .astype(str)
    .str.replace(',', '', regex=False)
)

metadata_clean['Cost'] = pd.to_numeric(
    metadata_clean['Cost'], errors='coerce'
)

print("Cost data type:", metadata_clean['Cost'].dtype)
print("\nCost summary:")
print(metadata_clean['Cost'].describe())



# Handle Missing Values in Restaurant Metadata

metadata_clean['Collections'] = metadata_clean['Collections'].fillna('Not Available')

metadata_clean['Timings'] = metadata_clean['Timings'].fillna('Not Available')

print("Missing values after treatment:")
print(metadata_clean[['Collections', 'Timings']].isnull().sum())

````

````python

# Inspect Rating Values

print("Unique rating values:")
print(reviews_clean['Rating'].value_counts(dropna=False))

print("\nNon-numeric rating values:")
numeric_ratings = pd.to_numeric(
    reviews_clean['Rating'], errors='coerce'
)

print(reviews_clean.loc[
    numeric_ratings.isna() & reviews_clean['Rating'].notna(),
    'Rating'
].unique())
````

````python

# Convert Rating Column into Numeric Format

reviews_clean['Rating'] = pd.to_numeric(
    reviews_clean['Rating'], errors='coerce'
)

print("Rating data type:", reviews_clean['Rating'].dtype)

print("\nRating distribution after conversion:")
print(reviews_clean['Rating'].value_counts(dropna=False).sort_index())

print("\nMissing ratings:", reviews_clean['Rating'].isnull().sum())



# Inspect Remaining Missing Values in Reviews

print("Missing values in each column:")
print(reviews_clean.isnull().sum())

print("\nRows containing missing values:")
display(reviews_clean[reviews_clean.isnull().any(axis=1)])

````

````python

# Remove Records with No Actual Review Information

reviews_clean = reviews_clean.dropna(
    subset=['Reviewer', 'Review', 'Rating', 'Metadata', 'Time'],
    how='all'
)

print("Shape after removing unusable records:", reviews_clean.shape)

print("\nRemaining missing values:")
print(reviews_clean.isnull().sum())
````

````python

# Convert Time Column into Datetime Format

reviews_clean['Time'] = pd.to_datetime(
    reviews_clean['Time'],
    errors='coerce'
)

print("Time data type:", reviews_clean['Time'].dtype)

print("\nTime range:")
print("Earliest:", reviews_clean['Time'].min())
print("Latest:", reviews_clean['Time'].max())

print("\nMissing timestamps:", reviews_clean['Time'].isnull().sum())
````

````python
# Extract reviewer reviews and followers

reviews_clean['Reviewer_Reviews'] = pd.to_numeric(
    reviews_clean['Metadata'].str.extract(r'(\d+)\s+Reviews?')[0],
    errors='coerce'
)

reviews_clean['Reviewer_Followers'] = pd.to_numeric(
    reviews_clean['Metadata'].str.extract(r'(\d+)\s+Followers?')[0],
    errors='coerce'
)

print(reviews_clean[
    ['Metadata', 'Reviewer_Reviews', 'Reviewer_Followers']
].head(10))

print("\nMissing values:")
print(reviews_clean[
    ['Reviewer_Reviews', 'Reviewer_Followers']
].isnull().sum())
````

````python
# Create indicators for missing reviewer information

reviews_clean['Reviewer_Reviews_Missing'] = (
    reviews_clean['Reviewer_Reviews'].isnull().astype(int)
)

reviews_clean['Reviewer_Followers_Missing'] = (
    reviews_clean['Reviewer_Followers'].isnull().astype(int)
)

# Fill missing values for numerical processing

reviews_clean['Reviewer_Reviews'] = (
    reviews_clean['Reviewer_Reviews'].fillna(0)
)

reviews_clean['Reviewer_Followers'] = (
    reviews_clean['Reviewer_Followers'].fillna(0)
)

print("Missing values after treatment:")
print(reviews_clean[
    ['Reviewer_Reviews', 'Reviewer_Followers']
].isnull().sum())

print("\nUpdated reviewer information:")
print(reviews_clean[
    ['Metadata', 'Reviewer_Reviews', 'Reviewer_Followers']
].head(10))
````

````python
# Check review text before cleaning

print("Review data type:", reviews_clean['Review'].dtype)

print("\nSample reviews:")
print(reviews_clean['Review'].dropna().head(10).to_string(index=False))

print("\nMissing reviews:", reviews_clean['Review'].isnull().sum())
````

````python
import re

# Create a separate cleaned review column

reviews_clean['Cleaned_Review'] = (
    reviews_clean['Review']
    .fillna('')
    .astype(str)
    .str.lower()
    .str.replace(r'\\n|\n', ' ', regex=True)
    .str.replace(r'[^a-z\s]', ' ', regex=True)
    .str.replace(r'\s+', ' ', regex=True)
    .str.strip()
)

# Display original and cleaned reviews

print(reviews_clean[
    ['Review', 'Cleaned_Review']
].head(10).to_string(index=False))

print("\nEmpty cleaned reviews:",
      (reviews_clean['Cleaned_Review'] == '').sum())
````

````python
# Inspect reviews that became empty after cleaning

empty_reviews = reviews_clean[
    reviews_clean['Cleaned_Review'] == ''
]

print("Number of empty cleaned reviews:", len(empty_reviews))

print("\nOriginal review values:")
print(empty_reviews['Review'].to_string())
````

````python
# Create a separate dataset for NLP sentiment analysis

sentiment_df = reviews_clean[
    reviews_clean['Cleaned_Review'].str.len() > 0
].copy()

print("Original reviews dataset:", reviews_clean.shape)
print("NLP sentiment dataset:", sentiment_df.shape)

print("\nEmpty reviews remaining:",
      (sentiment_df['Cleaned_Review'] == '').sum())

print("\nMissing cleaned reviews:",
      sentiment_df['Cleaned_Review'].isnull().sum())
````

````python
# Standardize restaurant names in both datasets

metadata_clean['Name'] = (
    metadata_clean['Name'].str.strip().str.lower()
)

reviews_clean['Restaurant'] = (
    reviews_clean['Restaurant'].str.strip().str.lower()
)

# Check unique restaurant counts

print("Unique restaurants in metadata:",
      metadata_clean['Name'].nunique())

print("Unique restaurants in reviews:",
      reviews_clean['Restaurant'].nunique())

# Check matching restaurant names

metadata_names = set(metadata_clean['Name'])
review_names = set(reviews_clean['Restaurant'])

print("\nMatching restaurants:",
      len(metadata_names.intersection(review_names)))

print("Restaurants only in metadata:",
      len(metadata_names - review_names))

print("Restaurants only in reviews:",
      len(review_names - metadata_names))
````

````python
# Identify restaurants without customer reviews

metadata_only = metadata_names - review_names

print("Restaurants present only in metadata:\n")

for restaurant in sorted(metadata_only):
    print(restaurant)
````

````python
# Final Data Quality Check

print("METADATA DATASET")
print("Shape:", metadata_clean.shape)
print("Duplicate rows:", metadata_clean.duplicated().sum())
print("\nMissing values:")
print(metadata_clean.isnull().sum())

print("\nREVIEWS DATASET")
print("Shape:", reviews_clean.shape)
print("Duplicate rows:", reviews_clean.duplicated().sum())
print("\nMissing values:")
print(reviews_clean.isnull().sum())

print("\nNLP DATASET")
print("Shape:", sentiment_df.shape)
print("Empty cleaned reviews:",
      (sentiment_df['Cleaned_Review'] == '').sum())
````

## ***5. Hypothesis Testing***

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

Answer Here.

### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

**Null Hypothesis (H₀):** There is no statistically significant relationship between restaurant cost and average customer rating.

**Alternative Hypothesis (H₁):** There is a statistically significant relationship between restaurant cost and average customer rating.

#### 2. Perform an appropriate statistical test.

````python

# Hypothesis 1: Pearson Correlation Test

from scipy.stats import pearsonr

# Select relevant columns and remove missing values
test_data = restaurant_analysis[
    ['Cost', 'Average_Rating']
].dropna()

# Perform Pearson correlation test
correlation, p_value = pearsonr(
    test_data['Cost'],
    test_data['Average_Rating']
)

print("Pearson Correlation Coefficient:", round(correlation, 4))
print("P-value:", p_value)

# Interpret the result
if p_value < 0.05:
    print("Result: Reject the Null Hypothesis (H0).")
    print("There is a statistically significant relationship.")
else:
    print("Result: Fail to Reject the Null Hypothesis (H0).")
    print("There is insufficient evidence of a significant relationship.")
````

The Pearson correlation test produced a correlation coefficient of **0.4213** and a p-value of **0.00001273**.

Since the p-value is less than the significance level of 0.05, we reject the null hypothesis.

This indicates a **moderate positive and statistically significant relationship** between restaurant cost and average customer rating. In this dataset, restaurants with higher costs tend to have higher average ratings. However, correlation does not imply causation.

##### Which statistical test have you done to obtain P-Value?

Pearson’s Correlation Coefficient (Pearson’s r) test was used to measure the strength and direction of the linear relationship between restaurant cost and average customer rating. The p-value was used to determine whether the observed correlation was statistically significant.

##### Why did you choose the specific statistical test?

Pearson’s correlation test was chosen because both variables—restaurant cost and average customer rating—are numerical. The test helps measure the strength and direction of their linear relationship and determine whether the correlation is statistically significant using the p-value.

### Hypothetical Statement - 2

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

H₀ (Null Hypothesis): There is no statistically significant correlation between customer ratings and TextBlob sentiment polarity scores.

H₁ (Alternative Hypothesis): There is a statistically significant correlation between customer ratings and TextBlob sentiment polarity scores.

#### 2. Perform an appropriate statistical test.

````python
from scipy.stats import spearmanr
from textblob import TextBlob

# Calculate TextBlob sentiment polarity
sentiment_df['TextBlob_Polarity'] = sentiment_df['Cleaned_Review'].apply(
    lambda text: TextBlob(text).sentiment.polarity
)

# Prepare data for statistical testing
test_data = sentiment_df[['Rating', 'TextBlob_Polarity']].dropna()

# Perform Spearman correlation test
correlation, p_value = spearmanr(
    test_data['Rating'],
    test_data['TextBlob_Polarity']
)

print("Spearman Correlation Coefficient:", correlation)
print("P-value:", p_value)

if p_value < 0.05:
    print("Result: Reject the Null Hypothesis (H0).")
    print("There is a statistically significant relationship.")
else:
    print("Result: Fail to Reject the Null Hypothesis (H0).")
    print("There is no statistically significant relationship.")
````

Spearman’s Rank Correlation Test.

##### Why did you choose the specific statistical test?

### Hypothetical Statement - 3

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

H₀ (Null Hypothesis): There is no statistically significant difference in customer ratings across positive, negative, and neutral sentiment categories.

H₁ (Alternative Hypothesis): There is a statistically significant difference in customer ratings across at least two sentiment categories.

#### 2. Perform an appropriate statistical test.

````python
from scipy.stats import kruskal

# Separate customer ratings by sentiment
positive_ratings = sentiment_df[
    sentiment_df['Sentiment'] == 'Positive'
]['Rating'].dropna()

negative_ratings = sentiment_df[
    sentiment_df['Sentiment'] == 'Negative'
]['Rating'].dropna()

neutral_ratings = sentiment_df[
    sentiment_df['Sentiment'] == 'Neutral'
]['Rating'].dropna()

# Perform Kruskal-Wallis test
statistic, p_value = kruskal(
    positive_ratings,
    negative_ratings,
    neutral_ratings
)

print("Kruskal-Wallis H-statistic:", statistic)
print("P-value:", p_value)

if p_value < 0.05:
    print("Result: Reject the Null Hypothesis (H0).")
    print("Customer ratings differ significantly across sentiment groups.")
else:
    print("Result: Fail to reject the Null Hypothesis (H0).")
    print("No statistically significant difference was found.")
````

##### Which statistical test have you done to obtain P-Value?

Kruskal-Wallis H Test was performed to compare customer ratings across three independent sentiment groups: Positive, Negative, and Neutral.

:##### Why did you choose the specific statistical test?

The Kruskal-Wallis test was chosen because we are comparing customer ratings across three independent sentiment groups. It is a non-parametric test that does not require the data to follow a normal distribution. It helps determine whether there are statistically significant differences in customer ratings among the three groups.
